```
open_source_levels = {
    "Fully Open": {
        "definition": "The model architecture, training code, and pre-trained weights are all publicly available.",
        "what_is_open": ["Model architecture", "Training code", "Pre-trained weights"],
        "what_you_can_do": ["Inspect internal workings", "Modify architecture/code", "Retrain from scratch", "Fine-tune"],
        "what_you_cannot_do": ["No limitations on modification or use if license permits"],
    },
    "Weights Released": {
        "definition": "Pre-trained model weights are publicly available, but the full training code or detailed architecture might be proprietary.",
        "what_is_open": ["Pre-trained weights"],
        "what_you_can_do": ["Fine-tune the model on new data", "Use for inference"],
        "what_you_cannot_do": ["Inspect or modify the original training process", "Retrain from scratch without proprietary code"],
    },
    "Architecture Only": {
        "definition": "Only the model's architectural design (e.g., research paper, network structure) is public, without pre-trained weights or training code.",
        "what_is_open": ["Model architecture/design principles"],
        "what_you_can_do": ["Understand the model's design", "Implement the architecture and train from scratch"],
        "what_you_cannot_do": ["Use for immediate inference", "Fine-tune a pre-trained model", "Access pre-trained knowledge"],
    },
}
open_source_levels
```

```
comparison_table = """| Openness level | What's open? | Impact on retraining/modifying |
| --- | --- | --- |
| Fully Open | Architecture, code, and weights | Full inspection, modification, and retraining possible |
| Weights Released | Pre-trained weights | Fine-tuning and domain adaptation possible, but not full training from scratch |
| Architecture Only | Model architecture | Only allows building and training a new model from scratch based on the design |"""
print(comparison_table)
```

```
comparative_paragraph = """Fully Open LLMs provide complete transparency, offering access to architecture, training code, and weights, which allows for deep customization and understanding. Weights Released models, while useful for fine-tuning and domain adaptation, limit insight into the original training process. In contrast, Architecture Only models merely provide the design, necessitating a full training effort from scratch to utilize them, thus offering the least immediate utility but maximum flexibility for novel implementations."""
print(comparative_paragraph)
```
