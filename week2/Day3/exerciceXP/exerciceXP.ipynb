```python
#exercice2

import numpy as np
from scipy import stats

# Jeu de données de l'exercice
data = [12, 15, 13, 12, 18, 20, 22, 21]

# 1. Calculs individuels
moyenne = np.mean(data)
mediane = np.median(data)
variance = np.var(data)      # ddof=1 pour la variance échantillonnale
ecart_type = np.std(data, ddof=1)    # ddof=1 pour l'écart-type échantillonnal

# 2. Utilisation de SciPy pour un résumé statistique complet
description = stats.describe(data)

# Affichage des résultats
print(f"--- Calculs de base ---")
print(f"Moyenne : {moyenne}")
print(f"Médiane : {mediane}")
print(f"Variance : {variance:.3f}")
print(f"Écart-type : {ecart_type:.3f}\n")

print(f"--- Résumé complet SciPy ---")
print(description)
```

```python
#exercice3

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# Paramètres de la distribution normale
moyenne = 50
ecart_type = 10

# 1. Générer des points sur l'axe X (couvrant 99.7% de la distribution : +/- 3 écarts-types)
x = np.linspace(moyenne - 3 * ecart_type, moyenne + 3 * ecart_type, 100)

# 2. Calculer la densité de probabilité (PDF) pour chaque point X
y = norm.pdf(x, loc=moyenne, scale=ecart_type)

# 3. Création du graphique
plt.figure(figsize=(8, 5))
plt.plot(x, y, label=f'Loi Normale ($\mu={moyenne}$, $\sigma={ecart_type}$)', color='blue', linewidth=2)
plt.fill_between(x, y, alpha=0.2, color='blue') # Remplissage sous la courbe

# Personnalisation du graphique
plt.title('Distribution Normale avec SciPy', fontsize=14)
plt.xlabel('Valeurs', fontsize=12)
plt.ylabel('Densité de probabilité', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()

# Affichage
plt.show()
```

```python
#exerice4

import numpy as np
from scipy import stats

# Génération des deux ensembles de données
np.random.seed(42) # Fixe la graine pour avoir des résultats reproductibles
data1 = np.random.normal(50, 10, 100)
data2 = np.random.normal(60, 10, 100)

# Application du test t de Student pour échantillons indépendants
t_stat, p_value = stats.ttest_ind(data1, data2)

# Affichage des résultats numériques
print(f"Statistique t : {t_stat:.4f}")
print(f"Valeur p (p-value) : {p_value:.4e}")

# Interprétation du résultat (seuil de signification de 5%)
alpha = 0.05
if p_value < alpha:
    print("\nConclusion : On rejette l'hypothèse nulle (H0).")
    print("Il y a une différence statistiquement significative entre les deux moyennes.")
else:
    print("\nConclusion : On ne peut pas rejeter l'hypothèse nulle (H0).")
    print("Il n'y a pas de différence statistiquement significative entre les deux moyennes.")
```

```python
#exerice5

import numpy as np
from scipy import stats

# Données de l'exercice
house_sizes = np.array([50, 70, 80, 100, 120])
house_prices = np.array([150000, 200000, 210000, 250000, 280000])

# Calcul de la régression linéaire
regression = stats.linregress(house_sizes, house_prices)

# Extraction des paramètres
pente = regression.slope
ordonnee = regression.intercept

# Estimation pour une maison de 90 m²
taille_cible = 90
prix_estime = pente * taille_cible + ordonnee

# Affichage des résultats
print(f"Pente (Slope) : {pente:.2f}")
print(f"Ordonnée à l'origine (Intercept) : {ordonnee:.2f}")
print(f"Prix estimé pour {taille_cible} m² : {prix_estime:.2f} unités monétaires")
```

```python
#exercie6

from scipy import stats

# Données de l'exercice
fertilizer_1 = [5, 6, 7, 6, 5]
fertilizer_2 = [7, 8, 7, 9, 8]
fertilizer_3 = [4, 5, 4, 3, 4]

# Application du test ANOVA à un facteur
f_stat, p_value = stats.f_oneway(fertilizer_1, fertilizer_2, fertilizer_3)

# Affichage des résultats numériques
print(f"Valeur F (F-statistic) : {f_stat:.4f}")
print(f"Valeur p (p-value) : {p_value:.4e}")
```

```python
#exercice7


from scipy.stats import binom

# Paramètres de l'exercice
n = 10       # Nombre de lancers (essais)
p = 0.5      # Probabilité de succès (obtenir face pour une pièce équilibrée)
k = 5        # Nombre exact de succès recherchés

# Calcul de la probabilité exacte (PMF)
probabilite = binom.pmf(k, n, p)

# Affichage du résultat
print(f"Probabilité d'obtenir exactement {k} faces : {probabilite:.4f}")
print(f"En pourcentage : {probabilite * 100:.2f}%")
```

```python
#exercice8

import pandas as pd
from scipy import stats

# Création du DataFrame d'exemple
data = pd.DataFrame({
    'age': [23, 25, 30, 35, 40], 
    'income': [35000, 40000, 50000, 60000, 70000]
})

# 1. Calcul du coefficient de corrélation de Pearson
pearson_coef, pearson_p = stats.pearsonr(data['age'], data['income'])

# 2. Calcul du coefficient de corrélation de Spearman
spearman_coef, spearman_p = stats.spearmanr(data['age'], data['income'])

# Affichage des résultats
print("--- Corrélation de Pearson ---")
print(f"Coefficient r : {pearson_coef:.4f}")
print(f"Valeur p      : {pearson_p:.4f}\n")

print("--- Corrélation de Spearman ---")
print(f"Coefficient rho : {spearman_coef:.4f}")
print(f"Valeur p        : {spearman_p:.4f}")
```
