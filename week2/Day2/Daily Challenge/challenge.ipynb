```python
# ==========================================
# Daily chalenge
# ==========================================
import pandas as pd
from sklearn.preprocessing import MinMaxScaler
from sklearn.decomposition import PCA

df = pd.read_csv(r"C:\Users\hp\Desktop\DI_BOOTCAMP\week2\Day2\datascience_salaries.csv")
print(df.head())

print("Aperçu du dataset :")
print(df.head())

# ==========================================
# 3. Normalisation Min-Max de la colonne salary
# ==========================================
scaler = MinMaxScaler()

df["salary_normalized"] = scaler.fit_transform(df[["salary"]])

print("\nColonne salary normalisée :")
print(df[["salary", "salary_normalized"]].head())


# 4. Réduction de dimensionnalité avec PCA

# Sélection des variables numériques
numeric_cols = df.select_dtypes(include=["int64", "float64"])

# PCA vers 2 composantes principales
pca = PCA(n_components=2)

pca_result = pca.fit_transform(numeric_cols)

# Création d'un nouveau DataFrame
df_pca = pd.DataFrame(
    pca_result,
    columns=["PC1", "PC2"]
)

print("\nDonnées après PCA :")
print(df_pca.head())

print("\nVariance expliquée :")
print(pca.explained_variance_ratio_)


salary_stats = (
    df.groupby("experience_level")["salary"]
      .agg(["mean", "median"])
      .reset_index()
)

print("\nSalaire moyen et médian par niveau d'expérience :")
print(salary_stats)


# 6. Remplacement des codes par des libellés

experience_mapping = {
    "EN": "Junior",
    "MI": "Intermédiaire",
    "SE": "Senior",
    "EX": "Executive"
}

salary_stats["experience_level"] = (
    salary_stats["experience_level"]
    .map(experience_mapping)
)

print("\nRésultat final :")
print(salary_stats)
```
