# Superstore Analysis - Final Corrected Version

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

from ipywidgets import interact, Dropdown, IntSlider
```

```python
df = pd.read_csv('Sample - Superstore.csv', encoding='latin1')

print(df.shape)
print(df.columns)

df.info()
print(df.describe())
print(df.isnull().sum())
```

```python
df['Order Date'] = pd.to_datetime(df['Order Date'])
df['Profit Margin'] = (df['Profit'] / df['Sales']) * 100
df['Order Month-Year'] = df['Order Date'].dt.to_period('M')
```

```python
monthly_sales = df.groupby(['Order Month-Year', 'Category'])['Sales'].sum().reset_index()
monthly_sales['Date'] = monthly_sales['Order Month-Year'].dt.to_timestamp()

def plot_sales(category):
    plt.figure(figsize=(12,6))

    if category == "All":
        data = df.groupby('Order Month-Year')['Sales'].sum()
        plt.plot(data.index.to_timestamp(), data.values, marker='o')
        plt.title("Monthly Sales Trend - All Categories")
    else:
        data = monthly_sales[monthly_sales['Category'] == category]
        plt.plot(data['Date'], data['Sales'], marker='o')
        plt.title(f"Monthly Sales Trend - {category}")

    plt.xticks(rotation=45)
    plt.grid()
    plt.show()

categories = ["All"] + list(df['Category'].unique())
interact(plot_sales, category=Dropdown(options=categories))
```

```python
state_sales = df.groupby('State')['Sales'].sum().sort_values()

def plot_top_states(top_n):
    plt.figure(figsize=(10,6))

    top_states = state_sales.tail(top_n)

    plt.barh(top_states.index, top_states.values)
    plt.title(f"Top {top_n} States by Sales")

    for i, v in enumerate(top_states.values):
        plt.text(v, i, f"${v:,.0f}")

    plt.show()

interact(plot_top_states, top_n=IntSlider(min=5, max=20, step=1, value=10))
```

```python
product_profit = df.groupby('Product Name')['Profit'].sum().sort_values(ascending=False).head(10)

plt.figure(figsize=(12,8))
ax = sns.barplot(x=product_profit.values, y=product_profit.index)

plt.title("Top 10 Most Profitable Products")

for i, v in enumerate(product_profit.values):
    ax.text(v, i, f"${v:,.0f}", va='center')

plt.show()
```

```python
total_sales = df['Sales'].sum()
total_profit = df['Profit'].sum()
profit_margin = (total_profit / total_sales) * 100

top_state = state_sales.index[-1]
top_product = product_profit.index[0]

print("=== EXECUTIVE SUMMARY ===\n")

print(f"Total Revenue: ${total_sales:,.0f}")
print(f"Total Profit: ${total_profit:,.0f}")
print(f"Profit Margin: {profit_margin:.2f}%\n")

print(f"Top State: {top_state}")
print(f"Top Product: {top_product}")
```
