1. Importance de la visualisation des données :
La visualisation des données permet de représenter les informations sous forme graphique afin de les rendre plus faciles à comprendre. Elle aide à identifier rapidement les tendances, les anomalies et les relations entre les données. Elle facilite également la communication des résultats et la prise de décision.

2. Utilité d'un graphique linéaire :
Un graphique linéaire sert à montrer l'évolution d'une variable au fil du temps. Il permet d'observer les tendances, les hausses et les baisses d'une donnée. Cet outil est très utile pour analyser les changements et prévoir les évolutions futures.

```python
import matplotlib.pyplot as plt

jours = ["Lun", "Mar", "Mer", "Jeu", "Ven", "Sam", "Dim"]
temperatures = [72, 74, 76, 80, 82, 78, 75]

plt.plot(jours, temperatures, marker='o')
plt.xlabel("Jour")
plt.ylabel("Température (°F)")
plt.title("Variation de la température sur une semaine")
plt.show()
```

```python
import matplotlib.pyplot as plt

mois = ["Jan", "Fév", "Mar", "Avr", "Mai"]
ventes = [5000, 5500, 6200, 7000, 7500]

plt.bar(mois, ventes)
plt.xlabel("Mois")
plt.ylabel("Montant des ventes ($)")
plt.title("Ventes mensuelles du magasin")
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("C:/Users/hp/Desktop/DI_BOOTCAMP/week2/day4/exerciceXP/Student Mental health.csv")

sns.histplot(data=df, x="What is your CGPA?", color="skyblue")

plt.title("Distribution des catégories de CGPA")
plt.xlabel("CGPA")
plt.ylabel("Nombre d'étudiants")

plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.countplot(
    data=df,
    x="Choose your gender",
    hue="Do you have Anxiety?",
    palette="Set2"
)

plt.title("Comparaison des niveaux d'anxiété selon le sexe")
plt.xlabel("Sexe")
plt.ylabel("Nombre d'étudiants")
plt.show()
```

```python
#exerice6

import seaborn as sns
import matplotlib.pyplot as plt

# Conversion des réponses en valeurs numériques
df["Panic_Attack_Num"] = df["Do you have Panic attack?"].map({
    "Yes": 1,
    "No": 0
})

# Création du nuage de points
sns.scatterplot(
    data=df,
    x="Age",
    y="Panic_Attack_Num",
    s=80,
    alpha=0.7
)

# Personnalisation
plt.title("Relation entre l'âge et les crises de panique")
plt.xlabel("Âge")
plt.ylabel("Crise de panique (0 = Non, 1 = Oui)")
plt.yticks([0, 1], ["Non", "Oui"])

# Affichage
plt.show()
```
