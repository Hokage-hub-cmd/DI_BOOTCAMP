```python
#exercice1
import numpy as np

# Création d'une matrice 3x3 inversible
A = np.array([[2, 1, 3], [0, 5, 6], [1, 0, 2]])

# 1. Calcul du déterminant
determinant = np.linalg.det(A)

# 2. Calcul de l'inverse (possible car le déterminant n'est pas nul)
inverse_A = np.linalg.inv(A)

print(f"Déterminant : {determinant}\nInverse :\n{inverse_A}")
```

```python
#exercice2

# Génération de 50 nombres aléatoires (distribution normale)
donnees = np.random.normal(loc=10, scale=2, size=50)

# Calculs statistiques
moyenne = np.mean(donnees)
mediane = np.median(donnees)
ecart_type = np.std(donnees)

print(f"Moyenne : {moyenne:.2f} | Médiane : {mediane:.2f} | Écart-type : {ecart_type:.2f}")
```

```python
#exercice5

import numpy as np
import matplotlib.pyplot as plt

# Création d'une matrice 5x5 valide avec des entiers (0 = noir, 255 = blanc)
# Cette matrice dessine une forme de croix blanche au centre
image_gris = np.array([
    [0,   0, 255,   0,   0],
    [0,   0, 255,   0,   0],
    [255, 255, 255, 255, 255],
    [0,   0, 255,   0,   0],
    [0,   0, 255,   0,   0]
])

print("Matrice NumPy de l'image 5x5 :")
print(image_gris)

# Affichage graphique de l'image en niveaux de gris
plt.imshow(image_gris, cmap='gray', vmin=0, vmax=255)
plt.title("Visualisation de la matrice 5x5")
plt.colorbar(label="Intensité (0=Noir, 255=Blanc)")
plt.show()
```

```python
#exercice 6

productivity_before = np.random.normal(loc=50, scale=10, size=30)
productivity_after = productivity_before + np.random.normal(loc=5, scale=3, size=30)

# Calcul de la différence de productivité pour chaque employé
differences = productivity_after - productivity_before

# Test de base : vérification de la moyenne des gains
gain_moyen = np.mean(differences)

# Si le gain moyen est nettement supérieur à 0, l'hypothèse est validée
print(f"Gain moyen de productivité : {gain_moyen:.2f}")
print("Résultat : Le programme est efficace" if gain_moyen > 0 else "Aucun effet")
```

```python
#exercice10 

import matplotlib.pyplot as plt

# Données aléatoires (marche aléatoire cumulée)
x = np.arange(100)
y = np.random.randn(100).cumsum()

# Graphique linéaire
plt.plot(x, y, color="purple", label="Évolution")
plt.title("Graphique linéaire de nombres aléatoires")
plt.xlabel("Index")
plt.ylabel("Valeurs cumulées")
plt.legend()
plt.show()
```
