```python
#exerice1

import numpy as np

# Création du tableau de 0 à 9
arr = np.arange(10)

print(arr)
```

```python
#exercice2
# Liste initiale
liste = [3.14, 2.17, 0, 1, 2]

# Conversion en tableau et changement de type en entier
arr = np.array(liste).astype(int)
print(arr)
```

```python
#exercice3

# Création d'une matrice 3x3 de 1 à 9
arr = np.arange(1, 10).reshape(3, 3)
print(arr)
```

```python
#exercice4
# Tableau de forme (4, 5) avec des nombres aléatoires
arr = np.random.rand(4, 5)
print(arr)
```

```python
#exercice5
# Tableau initial
arr = np.array([[21, 22, 23, 22, 22], [20, 21, 22, 23, 24], [21, 22, 23, 22, 22]])

# Sélection de la deuxième ligne (index 1)
seconde_ligne = arr[1]
print(seconde_ligne)
```

```python
#exercice6

# Tableau initial
arr = np.arange(10)

# Inversion des éléments
arr_inverse = arr[::-1]
print(arr_inverse)
```

```python
#exercice7
# Création d'une matrice identité 4x4
matrice_identite = np.eye(4)
print(matrice_identite)
```

```python
#exercice8
# Tableau initial
arr = np.arange(10)

# Calcul de la somme et de la moyenne
somme = arr.sum()
moyenne = arr.mean()

print(f"Sum: {somme}, Average: {moyenne}")
```

```python
#exercice9

# Création et restructuration en matrice 4x5
matrice = np.arange(1, 21).reshape(4, 5)
print(matrice)
```

```python
#exercice10

# Tableau initial de 0 à 9
arr = np.arange(10)

# Extraction des nombres impairs
impairs = arr[arr % 2 != 0]
print(impairs)
```
