```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Définition d'un seed pour obtenir les mêmes résultats à chaque exécution
np.random.seed(42)
```

```python
# 1. Listes pour les index et les colonnes
villes = ["Paris", "Abidjan", "Tokyo", "New York", "Londres", "Sydney", "Le Caire", "Montréal", "Moscou", "Rio"]
mois = ["Jan", "Fév", "Mar", "Avr", "Mai", "Juin", "Juil", "Août", "Sep", "Oct", "Nov", "Déc"]

# 2. Génération de la matrice NumPy (10 villes x 12 mois) entre -5 et 35°C
donnees_temperatures = np.random.uniform(-5, 35, size=(10, 12))

# 3. Création du DataFrame Pandas
df_meteo = pd.DataFrame(data=donnees_temperatures, index=villes, columns=mois)

# Affichage des premières lignes du DataFrame arrondi à 1 décimale
df_meteo.round(1)
```

```python
# 1. Calcul de la température moyenne annuelle pour chaque ville (axe horizontal)
moyennes_annuelles = df_meteo.mean(axis=1)

# 2. Identification des villes extrêmes
ville_chaude = moyennes_annuelles.idxmax()
ville_froide = moyennes_annuelles.idxmin()

# Affichage des résultats de l'analyse
print("--- TEMPÉRATURES MOYENNES ANNUELLES ---")
print(moyennes_annuelles.round(1).to_string())
print("\n--- SYNTHÈSE DES EXTRÊMES ---")
print(f"☀️ Ville la plus chaude : {ville_chaude} ({moyennes_annuelles[ville_chaude]:.1f}°C)")
print(f"❄️ Ville la plus froide : {ville_froide} ({moyennes_annuelles[ville_froide]:.1f}°C)")
```

```python
plt.figure(figsize=(12, 6))

# Tracer l'évolution mensuelle pour la ville la plus chaude et la plus froide
plt.plot(mois, df_meteo.loc[ville_chaude], marker='o', color='red', linewidth=2, label=f"Plus chaude : {ville_chaude}")
plt.plot(mois, df_meteo.loc[ville_froide], marker='s', color='blue', linewidth=2, label=f"Plus froide : {ville_froide}")

# Personnalisation du graphique
plt.title("Tendances mensuelles de température : Extrêmes Annuels", fontsize=14, fontweight='bold')
plt.xlabel("Mois", fontsize=12)
plt.ylabel("Température (°C)", fontsize=12)
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend(fontsize=11)

# Affichage du graphique
plt.show()
```

Ville la plus chaude : Montréal se distingue avec la température moyenne annuelle la plus élevée (17.6°C dans ce jeu de données synthétiques).
Ville la plus froide : Abidjan enregistre la température moyenne la plus basse (9.6°C)
.Observation sur la distribution : Comme les données ont été générées de manière uniforme (np.random.uniform), les fluctuations mensuelles ne suivent pas les saisons réelles de l'hémisphère Nord ou Sud. On observe des variations en dents de scie typiques de données purement aléatoires.
